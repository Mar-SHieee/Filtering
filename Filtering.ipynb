##**Feature Engineering**

```python
df["Gross_Sales"] = df["Unit_Price"] * df["Quantity"]
```

```python
df["Gross_Sales"]
```

```python
df["Net_Sales"] = df["Gross_Sales"] * (1 - df["Discount_Percent"] / 100)
```

```python
df["Net_Sales"]
```

```python
df["Total_Sales"]
```

###**Actually Net_Sales & Total_Sales are the same.**

```python
df["Profit_Margin"] = (df["Profit"] / df["Net_Sales"]) * 100
```

```python
df["Profit_Margin"]
```

```python
df["Profit"]
```

```python
df[["Gross_Sales", "Net_Sales", "Profit_Margin"]] = ( df[["Gross_Sales", "Net_Sales", "Profit_Margin"]].round(2))
```

```python
df[["Gross_Sales", "Net_Sales", "Profit_Margin"]].head()
```

## **Grouping and Aggregation**

```python
categories = (df.groupby("Product_Category")[["Total_Sales", "Profit"]].sum().sort_values(by="Profit", ascending=False))
```

```python
categories
```

##**Best Category : Furniture**

```python
best_category = (
    df.groupby("Product_Category")["Profit"]
      .sum()
      .idxmax()
)

best_category
```

```python
regions = (df.groupby("Region")[["Total_Sales", "Profit"]].sum().sort_values(by="Profit", ascending=False))
```

```python
regions
```

##**Strongest Region : Europe**

```python
strongest_region = (
    df.groupby("Region")["Profit"]
      .sum()
      .idxmax()
)

strongest_region
```

```python
Product_Name = (df.groupby("Product_Name")[["Total_Sales", "Profit"]].sum().sort_values(by="Profit", ascending=False))
```

```python
Product_Name
```

##**One Product To Promote is : Ergonomic Office Chair**

```python
best_product = (
    df.groupby("Product_Name")["Profit"]
      .sum()
      .idxmax()
)

best_product
```

## 📊 Final Business Insights

### ✅ Best Category
The **Furniture** category generated the highest total profit and total sales, making it the best-performing product category. This indicates strong customer demand and high revenue potential within the furniture segment.

### 🌍 Strongest Region
The **Europe** region achieved the highest overall profit, making it the strongest-performing region. This suggests that the company has a strong market presence and customer base in this region.

### 🚀 Product to Promote
The **Ergonomic Office Chair** generated the highest total profit among all products. Promoting this product through targeted marketing campaigns could further increase the company's profitability due to its proven financial performance.

### 💡 Additional Insights
- Office furniture products consistently rank among the top-performing products in both sales and profit.
- Some low-priced office supplies generated very low or even negative profit, indicating that their pricing strategy or operational costs should be reviewed.
- Focusing marketing efforts on high-profit products while optimizing underperforming products can improve overall business performance.
